```python
# Standard Library
import warnings

# Data Manipulation
import pandas as pd
import numpy as np

# Data Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Ignore warnings
warnings.filterwarnings("ignore")

# Visualization settings
plt.style.use("ggplot")
sns.set_theme(style="whitegrid")

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
df = pd.read_csv("../data/raw/train.csv")
```

```python
df_processed = df.copy()
```

```python
df_processed.drop(columns=["Loan_ID"], inplace=True)
```

```python
df_processed.isnull().sum()
```

```python
# Categorical
df_processed["Gender"] = df_processed["Gender"].fillna(df_processed["Gender"].mode()[0], inplace=True)
df_processed["Married"] = df_processed["Married"].fillna(df_processed["Married"].mode()[0], inplace=True)
df_processed["Dependents"] = df_processed["Dependents"].fillna(df_processed["Dependents"].mode()[0], inplace=True)
df_processed["Self_Employed"] = df_processed["Self_Employed"].fillna(df_processed["Self_Employed"].mode()[0], inplace=True)

# Numerical
df_processed["LoanAmount"] = df_processed["LoanAmount"].fillna(df_processed["LoanAmount"].median(), inplace=True)
df_processed["Loan_Amount_Term"] = df_processed["Loan_Amount_Term"].fillna(df_processed["Loan_Amount_Term"].median(), inplace=True)

# Binary
df_processed["Credit_History"] = df_processed["Credit_History"].fillna(df_processed["Credit_History"].mode()[0], inplace=True)
```

```python
df_processed.isnull().sum()
```

```python
df_processed["TotalIncome"] = (
    df_processed["ApplicantIncome"] +
    df_processed["CoapplicantIncome"]
)
```

```python
df_processed["LoanToIncomeRatio"] = (
    df_processed["LoanAmount"] /
    (df_processed["TotalIncome"] + 1)
)
```

```python
df_processed["Gender"] = df_processed["Gender"].map({"Male":1, "Female":0})

df_processed["Married"] = df_processed["Married"].map({"Yes":1, "No":0})

df_processed["Education"] = df_processed["Education"].map({
    "Graduate":1,
    "Not Graduate":0
})

df_processed["Self_Employed"] = df_processed["Self_Employed"].map({
    "Yes":1,
    "No":0
})

df_processed["Loan_Status"] = df_processed["Loan_Status"].map({
    "Y":1,
    "N":0
})
```

```python
df_processed["Dependents"] = (
    df_processed["Dependents"]
    .replace("3+", 3)
    .astype(int)
)
```

```python
df_processed = pd.get_dummies(
    df_processed,
    columns=["Property_Area"],
    drop_first=True
)
```
